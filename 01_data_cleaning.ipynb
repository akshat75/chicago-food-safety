# Phase 0: Data Cleaning & Preparation

**Goal:** Turn two raw, messy city datasets (~314K inspections + business licenses) into a clean, queryable relational database.

## Steps
1. Load raw CSVs → inspect schema, dtypes, nulls, duplicates
2. Standardize fields → dates, Risk category, Result, Facility Type
3. Reshape → explode packed `Violations` string into a normalized `violations` child table (one row per inspection-violation code)
4. Filter & dedupe licenses → food-specific license types only, latest record per business
5. Persist → write cleaned tables to `chicago_food_safety.db` (SQLite)

```python
import pandas as pd
```

```python
inspections=pd.read_csv('Food_Inspections.csv')
licenses=pd.read_csv('Business_Licenses.csv')
```

```python
inspections.head()
```

```python
licenses.head()
```

```python
inspections_clean=inspections[(inspections['License #'].notnull()) & (inspections['Risk'].notnull()) & (inspections['License #']!=0) & (inspections['Risk']!='All')].copy()
print(f"Before : {len(inspections)}")
print(f"After : {len(inspections_clean)}")
print(f"Dropped : {len(inspections)-len(inspections_clean)}")
```

```python
inspections_clean['License #']=inspections_clean['License #'].astype(int)
inspections_clean['Inspection Date']=pd.to_datetime(inspections_clean['Inspection Date'],format='%m/%d/%Y')
```

```python
inspections_clean.info()
```

```python
licenses['LICENSE DESCRIPTION'].value_counts(dropna=False)
```

```python
food_types = ['Retail Food Establishment', 'Wholesale Food Establishment', 'Special Event Food']
licenses_clean=licenses[licenses['LICENSE DESCRIPTION'].isin(food_types)].copy()
```

```python
licenses_clean['LICENSE TERM START DATE']=pd.to_datetime(licenses_clean['LICENSE TERM START DATE'],format='%m/%d/%Y', errors='coerce')
licenses_clean=licenses_clean.sort_values('LICENSE TERM START DATE', ascending=False).copy()
licenses_clean=licenses_clean.drop_duplicates(subset='LICENSE NUMBER', keep='first').copy()
licenses_clean.shape
```

```python
licenses_clean['LICENSE NUMBER']=pd.to_numeric(licenses_clean['LICENSE NUMBER'],errors='coerce')
licenses_clean=licenses_clean.dropna(subset='LICENSE NUMBER').copy()
licenses_clean['LICENSE NUMBER'] = licenses_clean['LICENSE NUMBER'].astype(int)
```

```python
inspections_clean.head(10)
```

```python
violations=inspections_clean[['Inspection ID','Violations']].dropna(subset='Violations').copy()
violations['Violations']=violations['Violations'].str.split(r' \| ')
violations_long=violations.explode('Violations')
print(violations_long.shape)
```

```python
violations_long.head(10)
```

```python
import re
extracted=violations_long['Violations'].str.extract(r'^(\d+)\.\s*(.*)')
violations_long['Violations_code']=extracted[0]
violations_long['Violations_text']=extracted[1]
print(violations_long['Violations_code'].isnull().sum())
violations_long.head(10)
```

```python
licenses_clean = licenses_clean[[
    'LICENSE NUMBER', 'LEGAL NAME', 'DOING BUSINESS AS NAME',
    'WARD', 'ZIP CODE', 'LICENSE DESCRIPTION', 'LICENSE STATUS'
]].copy()
print(licenses_clean.shape)
licenses_clean.head()
```

```python
violations_long[['Violations_description','Violations_comment']]=violations_long['Violations_text'].str.split(' - Comments:',n=1,expand=True)
violations_long.head(10)
```

```python
violations_clean=violations_long[['Inspection ID', 'Violations_code', 'Violations_description', 'Violations_comment']].copy()
violations_clean['Violations_code']=violations_clean['Violations_code'].astype(int)
violations_clean['Violations_description']=violations_clean['Violations_description'].str.strip()
violations_clean['Violations_comment']=violations_clean['Violations_comment'].str.strip()
```

```python
print(violations_clean.shape)
violations_clean.head()
```

```python
violations_clean['Violations_code'].value_counts().head(10)
```

```python
merged=pd.merge(inspections_clean,licenses_clean, left_on='License #', right_on='LICENSE NUMBER', how='left',indicator=True)
```

```python
print(merged['_merge'].value_counts())
```

```python
import sqlite3
```

```python
conn=sqlite3.connect('chicago_food_safety.db')
```

```python
inspections_clean.to_sql('Inspections',conn,if_exists='replace',index=False)
```

```python
licenses_clean.to_sql('licenses', conn, if_exists='replace', index=False)
violations_clean.to_sql('violations', conn, if_exists='replace', index=False)

conn.close()
print("done")
```
